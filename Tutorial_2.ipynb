```python
import numpy as np

# Create a 2x2 numpy array
array_2x2 = np.array([[1, 2], [3, 4]])
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4
random_array_3x4 = np.random.rand(3, 4)
print(random_array_3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3, 3))
zeroes_array = np.zeros((3, 3))

print("Ones array:\n", ones_array)
print("\nZeroes array:\n", zeroes_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
random_6x3 = np.random.rand(6, 3)
transposed_array = random_6x3.T

print("Original 6x3 array:\n", random_6x3)
print("\nTransposed array (3x6):\n", transposed_array)
```

```python
# 5. Create two random numpy arrays of size n x p and p x m and compute the dot product
n, p, m = 3, 4, 2

array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)
dot_product = np.dot(array_A, array_B)

print("Array A (n x p):\n", array_A)
print("\nArray B (p x m):\n", array_B)
print("\nDot Product (n x m):\n", dot_product)
```
