```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# 1. LOAD A SIMPLE DATASET
# -------------------------------------------------------------
# A simple dataset of student study habits and exam outcomes
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)

# Rename to make it ultra-simple and beginner-friendly
# Let's clean it slightly to keep only 5 basic columns
df = df[['species', 'island', 'bill_length_mm', 'flipper_length_mm', 'body_mass_g']].dropna().reset_index(drop=True)

print("="*60)
print("SIMPLE DATA SCIENCE ASSIGNMENT ANSWERS")
print("="*60)

# -------------------------------------------------------------
# 2, 3, 4, 5. SAMPLES, FEATURES, AND DATA TYPES
# -------------------------------------------------------------
num_rows, num_cols = df.shape
num_features = num_cols - 1  # Assuming last column (body_mass_g) or species is target

print(f"\n[Q1] First 5 rows of the dataset:")
print(df.head())

print(f"\n[Q3] Number of Features: {num_features}")
print(f"[Q4] Number of Samples/Observations: {num_rows}")

print("\n[Q2 & Q4 Evaluation]:")
print(f" -> Is {num_rows} samples enough? Yes, around {num_rows} rows is plenty for a simple analysis without being overwhelming.")
print(f" -> Is {num_features} features enough? Yes, 4 features are enough to see clear patterns without confusing the model.")

print("\n[Q5] Data Types in the Dataset:")
print(df.dtypes)

# -------------------------------------------------------------
# 6 & 8. MISSING / NULL VALUES
# -------------------------------------------------------------
null_counts = df.isnull().sum()

print("\n" + "="*60)
print("[Q6 & Q8] MISSING / NULL VALUES")
print("="*60)
print("Missing values per column:")
print(null_counts)
print("\nDoes it have missing values?")
print(" -> No, because we dropped empty rows to keep the dataset simple and clean.")

# -------------------------------------------------------------
# 7. FEATURES VS TARGET OUTPUT
# -------------------------------------------------------------
print("\n" + "="*60)
print("[Q7] FEATURES vs TARGET OUTPUT")
print("="*60)
print("Feature Columns: 'species', 'island', 'bill_length_mm', 'flipper_length_mm'")
print("Target Output Column: 'body_mass_g' (or 'species' if classifying species)")

# -------------------------------------------------------------
# 9. QUESTIONS TO ASK
# -------------------------------------------------------------
print("\n" + "="*60)
print("[Q9] QUESTIONS TO ASK & SUITABILITY")
print("="*60)
print("Question to Ask:")
print(" -> 'Can we predict a penguin's body weight (mass in grams) based on its flipper length and bill size?'")
print("\nDoes the dataset suit the question?")
print(" -> Yes, because physical measurements usually correlate strongly with body weight.")

# -------------------------------------------------------------
# VISUALIZATION (Simple & Easy to Explain)
# -------------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='flipper_length_mm', y='body_mass_g', hue='species')
plt.title('Flipper Length vs Body Mass')
plt.xlabel('Flipper Length (mm)')
plt.ylabel('Body Mass (g)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
